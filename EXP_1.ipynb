```python
import pandas as pd

df = pd.read_csv('/content/play_tennis.csv')

attributes = ['outlook', 'temp', 'humidity', 'wind']
hypothesis = None
count = 0

print("FIND-S Algorithm Steps:\n")

for _, row in df.iterrows():
    sample = [row[a] for a in attributes]

    if str(row['play']).strip().lower() == 'yes':
        count += 1

        if hypothesis is None:
            hypothesis = sample.copy()
            print(f"Example {count}: First positive example")
        else:
            for i in range(len(hypothesis)):
                if hypothesis[i] != sample[i]:
                    hypothesis[i] = '?'
            print(f"Example {count}: Positive example")

        print("Hypothesis:", hypothesis)

    else:
        print(f"Example: Negative example - Ignored")
        if hypothesis is not None:
            print("Hypothesis:", hypothesis)
    print()

print("Final Most Specific Hypothesis:")
print(hypothesis)

print("\nAttribute-wise Final Hypothesis:")
for attribute, value in zip(attributes, hypothesis):
    print(f"{attribute}: {value}")
```
